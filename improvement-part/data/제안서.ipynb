# 부산항 효율 제고 제안서

## 주제: 선석 통합운영을 통한 부산 신항 환적 체류시간 저감

---

## 요약

부산 신항의 환적 컨테이너는 야드에서 장시간 체류하고 있다. 본 제안서는 **"선박 대기시간을 줄이면 모선과 연결선의 연계가 빨라져 야드 체류시간도 감소한다"**는 가정을 세우고, 이를 선석 통합운영으로 실현하는 방안을 제시한다.

현재 신항 7개 터미널은 선석을 독립적으로 운영하여, 한 터미널에 입항이 몰려도 다른 터미널의 여유 선석을 쓰기 어렵다. 이에 기존 운영, 인접 터미널 선석 공유, 신항 전체 통합의 세 시나리오를 시뮬레이션으로 비교하고, 선박 대기시간·야드 체류시간·터미널 부하율의 변화로 가정의 성립 여부와 개선 가능성을 검증한다.

## 1. 현황 및 문제점

### 1) 현재 운영 방식

- **자부두 환적**: 한 터미널 안에서 양하 → 야드 장치 → 연결선 적하 일정에 맞춰 온샤시·반출.
- **타부두 환적**: 출발 터미널에서 반출한 뒤 다른 터미널로 이동하여 반입·장치·온샤시·반출 공정이 추가된다. 두 터미널의 야드 체류와 선박 일정을 함께 봐야 한다.
- **선석 운영**: PNC, PNIT, HPNT, HJNC, BNCT, BCT, DGT 7개 터미널이 각자 선석을 배정한다. 한 터미널에 입항이 집중되어도 다른 터미널의 빈 선석을 바로 쓰기 어렵다.

### 2) 분석에서 확인된 사실

| 구분 | 분석 결과 |
| :--- | :--- |
| 장기 체류 | HJNC 자부두 환적 1,490건의 평균 야드 체류 195.9시간, 중앙값 168.9시간. 120시간 이상이 1,093건(73.4%) |
| 방향별 차이 | HJNC–DGT 타부두 환적 38,196건에서 DGT→HJNC 평균 282.81시간, HJNC→DGT 평균 176.86시간.
| 선박 지표와의 관계 | 선박 시간 지표 5개와 야드 체류시간의 상관계수 절댓값이 모두 0.12 미만. 타부두·DGT 분석에서도 뚜렷한 선형 관계가 없음 |

※ 야드 체류시간은 야드 장치부터 온샤시까지의 시간이다. 타부두 환적은 두 터미널의 값을 합산한다.
※ 입항 지연시간, 작업시작전 대기시간, 작업시간, 작업 후 출항 대기시간, 출항 지연시간

### 3) 문제 해석: 왜 선석 배정을 바꿔 봐야 하는가

- 체류가 긴 컨테이너가 다수이므로 일부 극단 사례가 아닌 **구조적 체류**일 가능성이 있다. 다만 정상적인 연결선 대기가 포함되어 있어 체류 전체를 지연으로 보지는 않는다.
- 개별 선박의 시간 지표와 야드 체류시간의 선형 관계가 약하다는 것은 **선박 한 척의 지표로는 가정을 확인할 수 없다**는 뜻이며, 가정이 틀렸다는 증거로 보지 않는다.
- 체류시간은 모선과 연결선이 어떻게 맞물리는지에 좌우될 수 있다. 따라서 선석 배정을 바꿔 맞물림이 달라질 때의 변화를 **시뮬레이션으로 직접 비교**해야 한다.

### 4) 개선 필요성

- 개별 터미널 내부 대응에서 벗어나 신항 전체의 선석과 처리능력을 함께 활용하는 방안이 필요하다.
- 선석 통합운영이 선박 대기시간과 야드 체류시간을 줄이고, 과부하 터미널의 부담을 완화하는지 확인한다.

## 2. 개선 방안

### 1) 개선 방안 개요

7개 터미널의 선석을 공동으로 활용하는 **통합 배정 기준**을 마련하고, 이를 시뮬레이션으로 검증한다. 핵심은 한 터미널에 몰린 선박을 다른 터미널의 여유 선석으로 분산하되, 환적 연결 관계(모선–연결선)가 깨지지 않도록 하는 것이다.

### 2) 비교 시나리오

| 구분 | 운영방식 | 비교 목적 |
| :--- | :--- | :--- |
| 기존 운영 | 선박을 기존 터미널의 선석에 배정 | 비교 기준 |
| 인접 터미널 공유 | 기존 터미널과 인접 터미널의 선석을 공동 활용 | 제한적 공유의 효과 확인 |
| 신항 전체 통합 | 7개 터미널의 선석을 공동 활용 | 통합 범위 확대의 효과 확인 |

세 시나리오에는 동일한 입항 수요와 동일한 평가 기준을 적용한다.

### 3) 통합 배정 기준

1. **입력 자료 연계**: 7개 터미널의 선석배정현황과 선박 제원을 연계해 선박별 입항·접안·출항·선석 점유 정보를 정리한다.
2. **배정 가능 조건**: 선박 길이, 안벽 길이, 시간대별 점유 상태로 배정 가능한 선석을 구분한다. 수심·접안 능력·하역 설비는 자료 확보 범위에서 추가 반영한다.
3. **환적 연결 조건**: 모선과 연결선을 함께 고려하여 배정 변경이 연결 간격과 야드 체류시간에 미치는 영향을 본다. 한 선박의 접안이 빨라지는 효과만으로 전체 체류 감소를 판단하지 않는다. 필요 시 배정 우선순위와 최소 환적 소요시간을 조건으로 추가한다.
4. **부하율 평가**: 연간 하역능력과 처리 물량은 터미널별 부하율 평가에만 쓰며, 개별 선박의 시간당 작업속도와 동일하게 취급하지 않는다.

### 4) 시뮬레이션 검증 절차

1. 기존 운영을 재현하는 모델을 만들고 관측 자료와 비교해 재현 수준을 확인한다.
2. 세 시나리오에서 선박 대기시간, 야드 체류시간, 터미널별 부하율을 비교한다.
3. 반복 실험과 민감도 분석으로 선박 제원 결측, 배정 조건, 시간 가정이 결과에 미치는 영향을 확인한다.
4. 자료가 충분하면 기간을 나누어 검증하고, 결측 제원·선석 점유시간 보완이 필요하면 예측모델 활용을 검토한다.

### 5) 실행 시 고려사항

- 7개 터미널은 운영사가 다르므로 실제 적용에는 운영사, 선사, 항만 당국 간 협의가 필요하다.
- 장비 호환, 야드 연계, 상업적 이해관계가 걸림돌이 될 수 있다. 그래서 실제 운영 변경에 앞서 시뮬레이션으로 효과와 적용 조건을 먼저 확인한다.

## 3. 기대효과

### 1) 평가 지표

지표는 가정의 인과 순서(수단의 직접 효과 → 최종 목표 → 부작용 점검)대로 배치한다.

| 순서 | 지표 | 역할 | 산식·정의 |
| :--- | :--- | :--- | :--- |
| ① | 선박 대기시간 | 수단의 직접 효과 | 배정 가능 상태가 된 시점부터 접안까지의 시간. 평균 대기시간, 감소시간, 감소율로 평가. 실제 도착 자료가 부족하면 대체 기준을 명시 |
| ② | 야드 체류시간 | 최종 목표 | 야드 장치부터 온샤시까지의 시간(타부두는 두 터미널 합산). 감소율 = (기존 평균 − 개선 후 평균) ÷ 기존 평균 × 100 |
| ③ | 터미널별 부하율 | 부작용 점검 | 연간 처리 물량 ÷ 연간 하역능력 × 100. 최대 부하율과 터미널 간 편차로 평가 |

※ 물량을 분담하는 터미널은 부하율이 오를 수 있으므로, 모든 터미널의 부하율 감소가 아니라 **과부하 완화와 편차 축소**를 목표로 한다.

### 2) 성공 기준과 결과 해석

세 시나리오를 기존 운영과 동일 조건에서 비교하고, 결과에 따라 다음과 같이 해석한다. 목표 수치는 재현 모델의 결과와 현재 편차 수준을 확인한 뒤 확정한다.

| 시뮬레이션 결과 | 해석 | 후속 조치 |
| :--- | :--- | :--- |
| 선박 대기시간 ↓, 야드 체류시간 ↓ | **가정 성립**. 선석 통합이 체류 저감에 효과 있음 | 적용 가능 범위(인접 공유 / 전체 통합) 제시, 협의 사항 정리 |
| 선박 대기시간 ↓, 야드 체류시간 변화 없음 | 가정 불성립. 체류 원인은 연결선 일정, 장치 정책 등 다른 곳에 있음 | 연결선 일정·장치 정책 중심으로 원인 재탐색 |
| 대기시간·체류시간 ↓, 일부 터미널 부하율 과도 상승 | 효과는 있으나 부하 쏠림이 부작용 | 배정 우선순위·상한 조건을 추가해 재검증 |

어느 결과가 나와도 체류 저감 방안을 찾는 데 필요한 결론을 얻는다.

### 3) 정성적 효과

- **체류 관리**: 장기 체류 현황과 환적 연결 조건을 함께 파악해 체류 관리의 일관성 향상.
- **대기 관리**: 터미널별 여유 선석 정보를 공동 활용해 혼잡 시 대응 유연성 향상.
- **부하 관리**: 개별 터미널 중심 배정에서 항만 전체 자원 활용 중심으로 운영 기반 전환.

## 적용 범위 및 한계

- 환적 체류 분석 자료: HJNC·DGT 자부두 환적, HJNC↔DGT 타부두 환적 (2025년).
- 선석 운영 검토 대상: 7개 터미널 선석배정현황.
- 환적 표본의 결과를 신항 전체 화물의 효과로 확대하지 않는다.
- 실제 운영 변경 전에 과거 자료 기반의 가상 시나리오로만 검증한다.
- 지표별 시간 정의가 자료마다 다르면 비교 전에 기준을 통일한다.
